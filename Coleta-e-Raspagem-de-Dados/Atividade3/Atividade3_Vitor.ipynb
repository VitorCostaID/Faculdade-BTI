### **Questão 1**: Coleta e extração de dados com requests e BeautifulSoup

    
Considere a página de testes **Quotes to Scrape** (url: https://quotes.toscrape.com/), que disponibiliza frases, autores e informações sobre citações para fins educacionais.

Implemente um programa em Python que:

1. Utilize a biblioteca requests para realizar uma requisição HTTP à página.
2. Verifique se a requisição foi realizada com sucesso.
3. Utilize BeautifulSoup para analisar o HTML retornado.
4. Utilize o método find() para localizar e extrair:
    - o texto de uma citação;
    - o autor da citação;
    - identifique a tag HTML utilizada para apresentar o **texto** da citação.
5. Exiba os dados extraídos de forma organizada.







````python
import requests
from bs4 import BeautifulSoup
import time
import random
import logging
from datetime import datetime
````

## 1. Criando registros logging

````python
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger("scraper")
````

## 2. Funções de scraping

````python
path = "https://quotes.toscrape.com/"

# seu código

def requisicao_conteudo(url):
  response = requests.get(url)
  response.raise_for_status()

  return response.content

def parser_html(html):
  soup = BeautifulSoup(html, "html.parser")

  data = []

  for quote in soup.find_all(class_="quote"):
    data.append({
        "texto": quote.find(class_="text").get_text(strip=True),
        "autor": quote.find(class_="author").get_text(strip=True),
        "tags": [tag.get_text(strip=True) for tag in quote.find_all(class_="tag")]
    })

  return data

def main():
  logger.info(f"Acessando: {path}")

  # Fazer requisição do conteúdo do site
  response = requisicao_conteudo(path)
  quotes = parser_html(response)

  # Iterando sobre a lista de citações
  for quote in quotes:
      print(f'Citação: {quote["texto"]}')
      print(f'Autor: {quote["autor"]}')
      print(f"Tags: {', '.join(quote['tags'])}\n" + "-"*40)

  logger.info(f"{len(quotes)} registros extraídos")
````
